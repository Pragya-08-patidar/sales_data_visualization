```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

data = {
    "Order_ID": [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010,
                 1011,1012,1013,1014,1015,1016,1017,1018,1019,1020],

    "Order_Date": ["2026-01-05","2026-01-12","2026-01-20","2026-02-03","2026-02-15",
                   "2026-02-25","2026-03-04","2026-03-18","2026-03-28","2026-04-06",
                   "2026-04-19","2026-05-02","2026-05-15","2026-05-27","2026-06-05",
                   "2026-06-18","2026-07-03","2026-07-16","2026-08-08","2026-08-22"],

    "Product": ["Laptop","Mouse","Keyboard","Mobile","Headphones",
                "Laptop","Monitor","Mobile","Mouse","Keyboard",
                "Tablet","Laptop","Headphones","Monitor","Mobile",
                "Keyboard","Tablet","Laptop","Mouse","Headphones"],

    "Category": ["Electronics","Accessories","Accessories","Electronics","Accessories",
                 "Electronics","Electronics","Electronics","Accessories","Accessories",
                 "Electronics","Electronics","Accessories","Electronics","Electronics",
                 "Accessories","Electronics","Electronics","Accessories","Accessories"],

    "Region": ["North","South","East","West","North",
               "South","East","West","North","South",
               "East","West","North","South","East",
               "West","North","South","East","West"],

    "Quantity": [2,10,5,3,4,1,3,2,8,6,4,2,7,3,5,6,3,1,12,5],

    "Sales": [120000,15000,20000,75000,24000,60000,45000,50000,12000,18000,
              80000,110000,42000,48000,90000,21000,72000,65000,18000,30000],

    "Profit": [18000,4500,6000,9000,7000,12000,8000,7500,3500,5000,
               14000,16000,9000,8500,13000,6500,11000,10000,4500,8000]
}

df = pd.DataFrame(data)

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df
```

```python
total_sales = df["Sales"].sum()
print("Total Sales:", total_sales)
```

```python
total_quantity = df["Quantity"].sum()
print("Total Quantity Sold:", total_quantity)
```

```python
total_orders = df["Order_ID"].nunique()
print("Total Orders:", total_orders)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8,5))
plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

### Insight

This chart compares total sales generated by each product category.

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(8,5))
plt.bar(region_sales.index, region_sales.values)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

### Insight

The chart compares sales performance across different regions and helps identify the strongest-performing region.

```python
top_products = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(8,5))
plt.bar(top_products.index, top_products.values)

plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insight

The visualization identifies the products that generate the highest sales.

```python
df["Month"] = df["Order_Date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

### Insight

The monthly sales trend shows how sales performance changes over time and helps identify periods of high and low sales.

```python
monthly_profit = df.groupby("Month")["Profit"].sum()

plt.figure(figsize=(12,5))

plt.plot(
    monthly_profit.index.astype(str),
    monthly_profit.values,
    marker="o"
)

plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Sales"
)

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")

plt.show()
```

### Insight

The scatter plot helps analyze the relationship between the quantity sold and total sales.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Sales",
    bins=8,
    kde=True
)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

### Insight

The distribution shows how sales values are spread across the dataset.

```python
# Key Insights

print("Highest Sales Category:")
print(df.groupby("Category")["Sales"].sum().idxmax())

print("\nHighest Sales Region:")
print(df.groupby("Region")["Sales"].sum().idxmax())

print("\nTop Product by Sales:")
print(df.groupby("Product")["Sales"].sum().idxmax())

print("\nHighest Sales Month:")
print(df.groupby("Month")["Sales"].sum().idxmax())

print("\nHighest Profit Category:")
print(df.groupby("Category")["Profit"].sum().idxmax())
```

```python
import matplotlib.pyplot as plt

# KPI Values
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_quantity = df["Quantity"].sum()

# Create Dashboard
fig = plt.figure(figsize=(14, 10))

# KPI 1
ax1 = plt.subplot(3, 2, 1)
ax1.text(0.5, 0.5, f"Total Sales\n₹{total_sales:,.0f}",
         ha="center", va="center", fontsize=20)
ax1.axis("off")

# KPI 2
ax2 = plt.subplot(3, 2, 2)
ax2.text(0.5, 0.5, f"Total Profit\n₹{total_profit:,.0f}",
         ha="center", va="center", fontsize=20)
ax2.axis("off")

# KPI 3
ax3 = plt.subplot(3, 2, 3)
ax3.text(0.5, 0.5, f"Total Orders\n{total_orders}",
         ha="center", va="center", fontsize=20)
ax3.axis("off")

# KPI 4
ax4 = plt.subplot(3, 2, 4)
ax4.text(0.5, 0.5, f"Quantity Sold\n{total_quantity}",
         ha="center", va="center", fontsize=20)
ax4.axis("off")

# Sales by Region
ax5 = plt.subplot(3, 2, 5)
region_sales = df.groupby("Region")["Sales"].sum()
ax5.bar(region_sales.index, region_sales.values)
ax5.set_title("Sales by Region")
ax5.set_xlabel("Region")
ax5.set_ylabel("Sales")

# Sales by Category
ax6 = plt.subplot(3, 2, 6)
category_sales = df.groupby("Category")["Sales"].sum()
ax6.bar(category_sales.index, category_sales.values)
ax6.set_title("Sales by Category")
ax6.set_xlabel("Category")
ax6.set_ylabel("Sales")

plt.suptitle("Sales Data Analysis Dashboard", fontsize=22)
plt.tight_layout()
plt.show()
```
